```python
# ================================================================
# HORIZON-AWARE QUANTILE REGRESSOR (single model; raw policy sliders)
#   - Per-feature sign clamp caps (EWMs only; lags unconstrained)
#   - Minimal-invasive wiring for cap vector aligned to KNOWN_COLS
#   - (KEPT) lag_gate feature + dataset gating for lag features
#   - Caps tunable via Optuna at the "family" level (ewm6/ewm12 only)
#   - Validation mirrors serving: lag gate OFF
# ================================================================

# If running in Colab, uncomment:
!pip install -q "lightning==2.4.*"
!pip install -q "optuna==3.6.*" "optuna-integration==3.6.*"

import os, json, random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

import lightning.pytorch as pl
from lightning.pytorch.callbacks import EarlyStopping, ModelCheckpoint
from optuna.pruners import SuccessiveHalvingPruner
from optuna.integration import PyTorchLightningPruningCallback
import optuna

from google.colab import drive

# ---------------------------------------------------------------
# Repro & setup
# ---------------------------------------------------------------
SEED = 42
pl.seed_everything(SEED, workers=True)
np.random.seed(SEED); random.seed(SEED); torch.manual_seed(SEED)
torch.use_deterministic_algorithms(True, warn_only=True)

drive.mount("/content/drive")

# Paths
DATA_PATH = "/content/drive/MyDrive/Data/df_final_omicron.csv"
MODEL_DIR = "/content/drive/MyDrive/Models"
os.makedirs(MODEL_DIR, exist_ok=True)

# ---------------------------------------------------------------
# Load data
# ---------------------------------------------------------------
df = pd.read_csv(DATA_PATH)

# time alias used for centering
df["time"] = df["week_id"]

required_cols = {
    "country_iso3","year","week_id","iso_week","time",
    "covid_19_prevalence",
    "covid_19_policy_stringency","covid_19_face_covering_policy","covid_19_testing_tracing_policy",
    "latitude","sine_seasonality","cosine_seasonality",
}
missing = sorted(list(required_cols - set(df.columns)))
if missing:
    raise ValueError(f"Missing required columns: {missing}")

df["week_id"] = pd.to_numeric(df["week_id"], errors="raise")
df = df.sort_values(["country_iso3","week_id"]).reset_index(drop=True)

print("shape:", df.shape)
print("#countries:", df["country_iso3"].nunique())
print("years:", df["year"].min(), "→", df["year"].max())

# ---------------------------------------------------------------
# Train/val split by country
# ---------------------------------------------------------------
unique_countries = df["country_iso3"].unique().tolist()
random.shuffle(unique_countries)
k_val = max(1, int(0.2 * len(unique_countries)))
val_countries = set(unique_countries[:k_val])

df_train = df[~df["country_iso3"].isin(val_countries)].copy()
df_val   = df[df["country_iso3"].isin(val_countries)].copy()

# ---------------------------------------------------------------
# Policy sliders (raw; no demeaning)
# ---------------------------------------------------------------
POLICY_COLS = (
    "covid_19_policy_stringency",
    "covid_19_face_covering_policy",
    "covid_19_testing_tracing_policy",
)

# ---------------------------------------------------------------
# Koyck smoothing windows (built on RAW policy columns)
# ---------------------------------------------------------------
HALFLIVES = (6, 12)
EWM_TAGS  = [f"ewm_hl{hl}" for hl in HALFLIVES]

def build_policy_koyck(df_in: pd.DataFrame, group_col: str = "country_iso3") -> pd.DataFrame:
    df = df_in.copy()
    for hl in HALFLIVES:
        tag = f"ewm_hl{hl}"
        for feat in POLICY_COLS:
            sm = (
                df.groupby(group_col)[feat]
                  .apply(lambda s: s.ewm(halflife=hl, adjust=False, min_periods=1).mean())
                  .reset_index(level=0, drop=True)
            )
            df[f"{feat}_{tag}"] = sm.astype(float)
    return df

# ------------------ Explicit policy lags ------------------
LAG_STEPS = (1, 2, 3, 4)

def add_policy_lags(
    df_in: pd.DataFrame,
    policy_cols=POLICY_COLS,
    group_col: str = "country_iso3",
    lag_steps=LAG_STEPS
) -> pd.DataFrame:
    """Add simple discrete lags of raw policy sliders (per-country)."""
    df = df_in.sort_values([group_col, "week_id"]).copy()
    for feat in policy_cols:
        g = df.groupby(group_col)[feat]
        for k in lag_steps:
            col = f"{feat}_lag{k}"
            df[col] = g.shift(k).fillna(df[feat]).astype(float)
    return df

# ---------------------------------------------------------------
# Anchor features (history-only)
# ---------------------------------------------------------------
def add_anchor_history_features(
    df: pd.DataFrame,
    country_col: str = "country_iso3",
    time_col: str = "week_id",
    prevalence_col: str = "covid_19_prevalence",
    trend_window: int = 4,
) -> pd.DataFrame:
    df = df.sort_values([country_col, time_col]).copy()
    df[prevalence_col] = pd.to_numeric(df[prevalence_col], errors="coerce").fillna(0.0)

    # rolling mean (anchor)
    df["prev_ma_4w"] = (
        df.groupby(country_col)[prevalence_col]
          .rolling(window=trend_window, min_periods=1).mean()
          .reset_index(level=0, drop=True).astype(float)
    )

    # rolling slope (anchor)
    def _rolling_slope(series: pd.Series) -> float:
        y = series.to_numpy(dtype=float); k = y.shape[0]
        if k < 2: return 0.0
        x = np.arange(k, dtype=float); xm, ym = x.mean(), y.mean()
        denom = ((x - xm)**2).sum()
        if denom <= 0: return 0.0
        return float(((x - xm)*(y - ym)).sum() / denom)

    df["prev_slope_4w"] = (
        df.groupby(country_col)[prevalence_col]
          .rolling(window=trend_window, min_periods=2).apply(_rolling_slope, raw=False)
          .reset_index(level=0, drop=True).fillna(0.0).astype(float)
    )
    return df

# ---------------------------------------------------------------
# Synthetic countries (optional; default: enabled)
# ---------------------------------------------------------------
USE_SYNTH = True
SYNTH_BASE_BLEND_FEATURES = [
    "covid_19_prevalence",
    "covid_19_policy_stringency","covid_19_face_covering_policy","covid_19_testing_tracing_policy",
    "latitude",
]
SYNTH_PASSTHROUGH = ["week_id","year","iso_week","time","sine_seasonality","cosine_seasonality"]

def synthesize_countries(
    df_in: pd.DataFrame,
    country_col: str = "country_iso3",
    time_col: str = "week_id",
    ratios: tuple = (0.3, 0.5, 0.7),
    max_pairs: int = 26,
    same_hemisphere_only: bool = True,
    equator_band: float = 0.0,
    sample_with_replacement: bool = True,
    center_means: dict | None = None,
):
    countries = df_in[country_col].astype(str).unique().tolist()
    country_lat = (
        df_in[[country_col, "latitude"]]
        .dropna(subset=["latitude"]).drop_duplicates(subset=[country_col])
        .set_index(country_col)["latitude"]
    )
    countries = [c for c in countries if c in country_lat.index]
    def hemi_code(lat: float) -> int:
        lat = float(lat)
        return 0 if abs(lat) <= float(equator_band) else (1 if lat > 0.0 else -1)
    hemi = country_lat.apply(hemi_code)

    from itertools import combinations
    all_pairs = list(combinations(countries, 2))
    valid_pairs = [(c1,c2) for (c1,c2) in all_pairs if (hemi[c1] == hemi[c2])] if same_hemisphere_only else all_pairs
    if not valid_pairs:
        return pd.DataFrame(columns=[time_col, country_col] + SYNTH_BASE_BLEND_FEATURES + SYNTH_PASSTHROUGH)
    random.shuffle(valid_pairs)
    pairs = (valid_pairs[:max_pairs] if len(valid_pairs) >= max_pairs
             else (random.choices(valid_pairs, k=max_pairs) if sample_with_replacement else valid_pairs))

    numeric_cols = df_in.columns[df_in.dtypes.apply(lambda d: np.issubdtype(d, np.number))].tolist()
    frames = []
    for c1, c2 in pairs:
        s1 = df_in[df_in[country_col]==c1].set_index(time_col)
        s2 = df_in[df_in[country_col]==c2].set_index(time_col)
        idx = s1.index.intersection(s2.index)
        if len(idx)==0: continue
        s1, s2 = s1.loc[idx], s2.loc[idx]
        for r in ratios:
            out = pd.DataFrame(index=idx)
            for col in SYNTH_BASE_BLEND_FEATURES:
                if col in numeric_cols: out[col] = (1-r)*s1[col].astype(float) + r*s2[col].astype(float)
            for pc in SYNTH_PASSTHROUGH:
                if pc in s1.columns: out[pc] = s1[pc].values
            out = out.reset_index(); out[country_col] = f"{c1}_{c2}_mix_{int(r*100)}"
            frames.append(out)

    if not frames:
        return pd.DataFrame(columns=[time_col, country_col] + SYNTH_BASE_BLEND_FEATURES + SYNTH_PASSTHROUGH)

    df_blend = (pd.concat(frames, ignore_index=True)
                .sort_values([country_col, time_col]).reset_index(drop=True))
    if time_col in df_blend and pd.api.types.is_float_dtype(df_blend[time_col]): df_blend[time_col] = df_blend[time_col].round().astype(int)
    if "year" in df_blend and pd.api.types.is_float_dtype(df_blend["year"]): df_blend["year"] = df_blend["year"].round().astype(int)

    # Build windows + anchors BEFORE centering
    df_blend = build_policy_koyck(df_blend, group_col=country_col)
    df_blend = add_policy_lags(df_blend, group_col=country_col)
    df_blend = add_anchor_history_features(df_blend, country_col=country_col, time_col=time_col,
                                           prevalence_col="covid_19_prevalence", trend_window=4)

    # Center moderators (TRAIN means) — only after anchors exist
    CENTER_KEYS = ["prev_ma_4w","prev_slope_4w","time"]
    for col in CENTER_KEYS:
        if col not in df_blend.columns: continue
        mu = float(center_means[col]) if (center_means is not None and col in center_means) else float(df_blend[col].mean())
        df_blend[f"{col}_c"] = df_blend[col].astype(float) - mu

    return df_blend

# ---------------------------------------------------------------
# Build train/val features
# ---------------------------------------------------------------
df_train_feat = build_policy_koyck(df_train, group_col="country_iso3")
df_val_feat   = build_policy_koyck(df_val,   group_col="country_iso3")

df_train_feat = add_policy_lags(df_train_feat, group_col="country_iso3")
df_val_feat   = add_policy_lags(df_val_feat,   group_col="country_iso3")

df_train_feat = add_anchor_history_features(df_train_feat, country_col="country_iso3", time_col="week_id",
                                            prevalence_col="covid_19_prevalence", trend_window=4)
df_val_feat   = add_anchor_history_features(df_val_feat,   country_col="country_iso3", time_col="week_id",
                                            prevalence_col="covid_19_prevalence", trend_window=4)

# Center moderators AFTER anchors exist; only what's used
CENTER_FOR_INTS = ["prev_ma_4w","prev_slope_4w","time"]
CENTER_MEANS_FOR_INTS = df_train_feat[CENTER_FOR_INTS].astype(float).mean().to_dict()
for col in CENTER_FOR_INTS:
    mu = float(CENTER_MEANS_FOR_INTS[col])
    df_train_feat[f"{col}_c"] = df_train_feat[col].astype(float) - mu
    df_val_feat[f"{col}_c"]   = df_val_feat[col].astype(float)   - mu

# Interactions are currently OFF (no *_relu moderators in use)

# Seasonality × latitude interactions (used)
def build_interactions(df_in: pd.DataFrame) -> pd.DataFrame:
    df = df_in.copy()
    df["sine_seasonality_x_latitude"]   = df["sine_seasonality"].astype(float) * df["latitude"].astype(float)
    df["cosine_seasonality_x_latitude"] = df["cosine_seasonality"].astype(float) * df["latitude"].astype(float)
    return df

df_train_feat = build_interactions(df_train_feat)
df_val_feat   = build_interactions(df_val_feat)

# Optional synthetic augmentation
if USE_SYNTH:
    df_synth = synthesize_countries(
        df_in=df_train,
        country_col="country_iso3",
        time_col="week_id",
        center_means=CENTER_MEANS_FOR_INTS
    )
    if len(df_synth):
        df_synth = build_interactions(df_synth)
        df_synth = df_synth[df_synth["year"] > 2019].copy()
        df_train_feat = pd.concat([df_train_feat, df_synth], ignore_index=True)

# Focus on pandemic years
df_train_feat = df_train_feat[df_train_feat["year"] > 2019].copy()
df_val_feat   = df_val_feat[df_val_feat["year"] > 2019].copy()

# ---------------------------------------------------------------
# Horizon supervision
# ---------------------------------------------------------------
H = 12

def _policy_base_names():
    base = [f"{feat}_{tag}" for tag in EWM_TAGS for feat in POLICY_COLS]
    return base, []  # (policy EWMs, interactions OFF)

policy_base, policy_allmods = _policy_base_names()

# Lag feature basenames
POLICY_LAG_BASENAMES = [f"{feat}_lag{k}" for feat in POLICY_COLS for k in LAG_STEPS]

# (NEW) handy name list for lag columns
LAG_FEATURES = [f"{feat}_lag{k}" for feat in POLICY_COLS for k in LAG_STEPS]

# KNOWN_COLS … keep as-is but append lag_gate at the end
KNOWN_COLS = [
    # anchors / time (centered where *_c is explicit)
    "prev_ma_4w_c","prev_slope_4w_c","time_c",
    # geography & seasonality
    "latitude","sine_seasonality","cosine_seasonality",
    "sine_seasonality_x_latitude","cosine_seasonality_x_latitude",
    # policies: discrete lags + Koyck windows
    *POLICY_LAG_BASENAMES,     # lags
    *policy_base, *policy_allmods,
    # (NEW)
    "lag_gate",
]

def build_horizon_supervision(
    df_in: pd.DataFrame, country_col="country_iso3", time_col="week_id",
    target_col="covid_19_prevalence", H=8, c2i: dict[str,int] | None = None,
):
    df = df_in.copy()
    if c2i is None:
        countries = sorted(df[country_col].astype(str).unique().tolist())
        c2i = {c: i for i, c in enumerate(countries)}
    df["country_idx"] = df[country_col].astype(str).map(c2i).astype(int)
    for h in range(1, H+1):
        df[f"y_h{h}"] = df.groupby(country_col)[target_col].shift(-h)
    keep = np.ones(len(df), dtype=bool)
    for h in range(1, H+1):
        keep &= df[f"y_h{h}"].notna().values
    return df.loc[keep].reset_index(drop=True), c2i

all_countries = sorted(set(df_train_feat["country_iso3"].astype(str)) | set(df_val_feat["country_iso3"].astype(str)))
country_index_map = {c: i for i, c in enumerate(all_countries)}

df_train_sup, _ = build_horizon_supervision(df_train_feat, H=H, c2i=country_index_map)
df_val_sup,   _ = build_horizon_supervision(df_val_feat,   H=H, c2i=country_index_map)

# ---------------------------------------------------------------
# Feature normalization (train-only)
# ---------------------------------------------------------------
FEATURE_NORM_JSON = os.path.join(MODEL_DIR, "feature_norm_stats_omicron.json")

for col in KNOWN_COLS:
    if col not in df_train_feat.columns: df_train_feat[col] = 0.0
    if col not in df_val_feat.columns:   df_val_feat[col]   = 0.0

_train_X = df_train_feat[KNOWN_COLS].astype(float)
train_means = _train_X.mean(axis=0)
train_stds  = _train_X.std(axis=0).replace(0.0, 1.0)

df_train_feat[KNOWN_COLS] = (df_train_feat[KNOWN_COLS].astype(float) - train_means) / train_stds
df_val_feat[KNOWN_COLS]   = (df_val_feat[KNOWN_COLS].astype(float)   - train_means) / train_stds

feature_norm = {col: {"mean": float(train_means[col]), "std": float(train_stds[col])} for col in KNOWN_COLS}

with open(FEATURE_NORM_JSON, "w") as f: json.dump(feature_norm, f, indent=2)
print(f"✅ Feature normalization written → {FEATURE_NORM_JSON}")

df_train_sup, _ = build_horizon_supervision(df_train_feat, H=H, c2i=country_index_map)
df_val_sup,   _ = build_horizon_supervision(df_val_feat,   H=H, c2i=country_index_map)

# ---------------------------------------------------------------
# Targets: stabilized logit on y/U with cap scaling
# ---------------------------------------------------------------
TARGET_CAP_U = 1.00
EPS_LOGIT = 1e-6

def _logit01(y: np.ndarray, eps: float = EPS_LOGIT) -> np.ndarray:
    y = np.asarray(y, dtype=np.float64)
    y = np.clip(y, eps, 1.0 - eps)
    return np.log(y) - np.log1p(-y)

def make_logit_targets_scaled(df_in: pd.DataFrame, H: int, U: float, target_prefix_out: str = "y_h"):
    df = df_in.copy()
    invU = 1.0 / float(U)
    for h in range(1, H+1):
        col = f"{target_prefix_out}{h}"
        y = df[col].to_numpy(dtype=float) * invU
        z = _logit01(y, eps=EPS_LOGIT)
        df[f"{col}_logit"] = z.astype(np.float32)
    return df

df_train_sup = make_logit_targets_scaled(df_train_sup, H=H, U=TARGET_CAP_U)
df_val_sup   = make_logit_targets_scaled(df_val_sup,   H=H, U=TARGET_CAP_U)

with open(FEATURE_NORM_JSON, "r") as _f:
    NORM_STATS = json.load(_f)

# ---------------------------------------------------------------
# Torch datasets & loaders
# ---------------------------------------------------------------
from torch.utils.data import Dataset, DataLoader
QUANTILES = [0.1, 0.5, 0.9]

# STEP_VARY_COLS: features that update over horizon
SEASONAL_FEATURES = [
    "sine_seasonality","cosine_seasonality",
    "sine_seasonality_x_latitude","cosine_seasonality_x_latitude",
]
STEP_VARY_COLS = set([
    *[f"{feat}_{tag}" for tag in EWM_TAGS for feat in POLICY_COLS],  # Koyck windows
    *[f"{feat}_lag{k}" for feat in POLICY_COLS for k in LAG_STEPS],  # discrete lags
    *SEASONAL_FEATURES,
])

# (interactions are OFF; function kept for compatibility if later enabled)
INT_MODS: list[str] = []

def _recompute_policy_interactions_inplace(xh: dict, x_anchor: dict, norm_stats: dict):
    # No-op when INT_MODS is empty; kept for drop-in compatibility.
    return

class HorizonDataset(Dataset):
    def __init__(self, df_in, known_cols, H, country_col="country_iso3", time_col="week_id",
                 norm_stats: dict | None = None,
                 gate_mode: str = "fixed",   # "sample" or "fixed"
                 gate_p: float = 0.5,        # only used if gate_mode="sample"
                 gate_value: int = 1):       # only used if gate_mode="fixed"
        self.df = df_in.reset_index(drop=True)
        self.known_cols = known_cols
        self.H = H
        self.country_col = country_col
        self.time_col = time_col
        self.norm_stats = norm_stats or {}
        self.gate_mode  = gate_mode
        self.gate_p     = float(gate_p)
        self.gate_value = int(gate_value)

        cols_needed = list(set(known_cols) | {country_col, time_col, *INT_MODS})
        slim = self.df[cols_needed].copy()
        self._by_key = {(str(r[country_col]), int(r[time_col])): r for _, r in slim.iterrows()}

        # cache indices of lag features in KNOWN_COLS for fast gating
        self.lag_idx = [i for i, n in enumerate(self.known_cols) if n in LAG_FEATURES]
        self.gate_idx = self.known_cols.index("lag_gate")

    def __len__(self):
        return len(self.df)

    def _sample_gate(self):
        if self.gate_mode == "sample":
            return 1.0 if (np.random.rand() < self.gate_p) else 0.0
        return float(self.gate_value)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        ctry = str(row[self.country_col])
        t0   = int(row[self.time_col])

        # base (already Z-scored in the dataframe)
        x0 = {c: float(row[c]) if c in row and pd.notna(row[c]) else 0.0 for c in self.known_cols}

        # choose the gate for this sample
        gate = self._sample_gate()
        x0["lag_gate"] = gate  # in Z-space; 0/1 fine after Z-norm

        # Anchor values for interactions (inactive in your config, kept for completeness)
        x_anchor = {m: (float(row[m]) if (m in row and pd.notna(row[m])) else 0.0) for m in INT_MODS}
        x_anchor.update({c: x0.get(c, 0.0) for c in self.known_cols})

        Xh = np.zeros((self.H, len(self.known_cols)), dtype=np.float32)
        for h in range(1, self.H+1):
            xh = dict(x0)
            fut = self._by_key.get((ctry, t0 + h), None)
            if fut is not None:
                for col in STEP_VARY_COLS:
                    if col in self.known_cols and col in fut and pd.notna(fut[col]):
                        xh[col] = float(fut[col])

            _recompute_policy_interactions_inplace(xh, x_anchor, self.norm_stats)

            # apply the gate to all lag features (multiply in Z-space)
            if self.lag_idx:
                for i in self.lag_idx:
                    n = self.known_cols[i]
                    xh[n] = float(xh.get(n, 0.0)) * gate

            # write row in KNOWN_COLS order
            Xh[h-1, :] = np.fromiter((xh.get(c, 0.0) for c in self.known_cols),
                                     dtype=np.float32, count=len(self.known_cols))

        y = np.stack([row[f"y_h{h}_logit"] for h in range(1, self.H+1)], axis=0).astype(np.float32)
        country_idx = int(row["country_idx"])
        return torch.from_numpy(Xh), torch.from_numpy(y), torch.tensor(country_idx, dtype=torch.long)

# Create loaders with gating settings
train_dataset = HorizonDataset(
    df_train_sup, KNOWN_COLS, H, norm_stats=NORM_STATS,
    gate_mode="sample", gate_p=0.5   # mix of on/off during training
)

# ✅ Validation mirrors serving (lag gate OFF)
val_dataset   = HorizonDataset(
    df_val_sup, KNOWN_COLS, H, norm_stats=NORM_STATS,
    gate_mode="fixed", gate_value=0
)

batch_size = 1024
train_loader = DataLoader(train_dataset,
                          batch_size=batch_size,
                          shuffle=True,
                          num_workers=4,
                          pin_memory=True,
                          persistent_workers=True,
                          prefetch_factor=2,
                          drop_last=False)
val_loader   = DataLoader(val_dataset,
                          batch_size=batch_size*2,
                          shuffle=False,
                          num_workers=4,
                          pin_memory=True,
                          persistent_workers=True,
                          prefetch_factor=2,
                          drop_last=False)

# ---------------------------------------------------------------
# Per-feature negative caps for policy EWMs ONLY (lags unconstrained)
# ---------------------------------------------------------------
# Defaults for EWMs only (used as fallbacks)
DEFAULT_CAPS_BY_SUFFIX = {
    "_ewm_hl6":  0.10,
    "_ewm_hl12": 0.12,
}

# (Optional) name->cap map (only EWMs). Leave empty to rely on suffix defaults.
POLICY_CAPS_BY_NAME = {}
for n in KNOWN_COLS:
    if n.startswith("covid_19_") and ("_ewm_hl" in n):
        for suf, cap in DEFAULT_CAPS_BY_SUFFIX.items():
            if n.endswith(suf):
                POLICY_CAPS_BY_NAME[n] = cap
                break

# --- keep negative caps ONLY for policy EWMs; lags are UNCONSTRAINED ---
POLICY_NEG_IDX = []
POLICY_CAPS_VEC = []
for i, n in enumerate(KNOWN_COLS):
    is_policy = n.startswith("covid_19_")
    is_ewm    = ("_ewm_hl" in n)
    # NOTE: lags intentionally excluded from clamp
    if is_policy and is_ewm:
        POLICY_NEG_IDX.append(i)
        # reuse your per-family map or fallback cap
        cap = (
            POLICY_CAPS_BY_NAME.get(n, None) if n in POLICY_CAPS_BY_NAME else
            (0.10 if n.endswith("_ewm_hl6")  else
             0.12 if n.endswith("_ewm_hl12") else 0.15)
        )
        POLICY_CAPS_VEC.append(cap)

# ---------------------------------------------------------------
# Seasonality (mains + their interaction with latitude)
# ---------------------------------------------------------------
SEASONAL_FEATURES = [
    "sine_seasonality","cosine_seasonality",
    "sine_seasonality_x_latitude","cosine_seasonality_x_latitude",
]

# ---------------------------------------------------------------
# Model (single model)
# ---------------------------------------------------------------
class HorizonQuantileModel(nn.Module):
    def __init__(self,
                 d_in: int,
                 H: int,
                 q_list=(0.1, 0.5, 0.9),
                 horz_dim: int = 8,
                 hidden: int | None = 128,
                 dropout: float = 0.0,
                 enable_monotone_hidden: bool = True,
                 # vectorized caps for selected policy features
                 policy_feature_indices_neg=None,
                 policy_caps_vec=None,
                 # backward-compat aliases (optional)
                 policy_idx_neg: torch.Tensor | None = None,
                 policy_caps_neg: torch.Tensor | None = None):
        super().__init__()
        self.d_in = d_in
        self.H = H
        self.q_list = list(q_list)
        self.Q = len(self.q_list)

        self.h_emb = nn.Embedding(H, horz_dim)

        self.fe_in = d_in + horz_dim
        self.use_hidden = hidden is not None
        self.dropout = nn.Dropout(dropout) if (self.use_hidden and dropout > 0) else nn.Identity()

        if not self.use_hidden:
            self.fixed_head  = nn.Linear(self.fe_in, 1, bias=False)
            self.quantile_out = nn.Linear(1, self.Q, bias=True)
        else:
            self.fixed_head  = nn.Linear(self.fe_in, hidden, bias=True)
            self.quantile_out = nn.Linear(hidden, self.Q, bias=True)

        # Monotone path (sign clamps on specified policy features)
        self.enable_monotone_hidden = bool(enable_monotone_hidden and self.use_hidden)

        if self.enable_monotone_hidden:
            W0 = self.fixed_head.weight.detach().clone()
            self.fixed_head_weight_free = nn.Parameter(W0)
            self.fixed_head_bias_param = self.fixed_head.bias

            # prefer new args; fall back to old if provided
            neg = policy_feature_indices_neg if policy_feature_indices_neg is not None else policy_idx_neg
            caps = policy_caps_vec if policy_caps_vec is not None else policy_caps_neg
            neg = neg if neg is not None else []
            caps = caps if caps is not None else [0.15] * (len(neg) if hasattr(neg, "__len__") else 0)

            # ensure tensors
            if isinstance(neg, torch.Tensor):
                neg_t = neg.to(dtype=torch.long)
            else:
                neg_t = torch.tensor(neg, dtype=torch.long)
            if isinstance(caps, torch.Tensor):
                caps_t = caps.to(dtype=torch.float32)
            else:
                caps_t = torch.tensor(caps, dtype=torch.float32)

            self.register_buffer("_policy_idx_neg", neg_t)
            self.register_buffer("_policy_caps", caps_t)

            qW0 = self.quantile_out.weight.detach().clone()
            self.quantile_out_weight_free = nn.Parameter(qW0)
            self.quantile_out_bias_param = self.quantile_out.bias

    @staticmethod
    def _non_crossing(q_raw: torch.Tensor) -> torch.Tensor:
        q10 = q_raw[..., 0:1]
        d2  = F.softplus(q_raw[..., 1:2])
        d3  = F.softplus(q_raw[..., 2:3])
        q50 = q10 + d2
        q90 = q50 + d3
        return torch.cat([q10, q50, q90], dim=-1)

    def forward(self, decoder_cont: torch.Tensor) -> torch.Tensor:
        B, H, D = decoder_cont.shape
        assert H == self.H and D == self.d_in

        h_ids = torch.arange(self.H, device=decoder_cont.device).view(1, H).expand(B, H)
        h_vec = self.h_emb(h_ids)

        fe = torch.cat([decoder_cont, h_vec], dim=-1)  # [B,H,fe_in]

        if self.enable_monotone_hidden:
            W_free = self.fixed_head_weight_free
            eff_W = W_free.clone()

            # per-index caps for all selected policy features (EWMs only)
            if self._policy_idx_neg.numel() > 0:
                sel  = eff_W[:, self._policy_idx_neg]             # [hidden, K]
                caps = self._policy_caps.view(1, -1)              # [1, K]
                eff_W[:, self._policy_idx_neg] = -(torch.sigmoid(sel) * caps)

            h1 = fe @ eff_W.t()
            if self.fixed_head_bias_param is not None:
                h1 = h1 + self.fixed_head_bias_param.view(1, 1, -1)
            h1 = F.relu(h1)
            h1 = self.dropout(h1)

            qW = F.softplus(self.quantile_out_weight_free)
            out_raw = h1 @ qW.t()
            if self.quantile_out_bias_param is not None:
                out_raw = out_raw + self.quantile_out_bias_param.view(1, 1, -1)
            return self._non_crossing(out_raw)

        if not self.use_hidden:
            mu = self.fixed_head(fe)
            out_raw = self.quantile_out(mu)
            return self._non_crossing(out_raw)
        else:
            h1 = self.fixed_head(fe)
            h1 = F.relu(h1); h1 = self.dropout(h1)
            out_raw = self.quantile_out(h1)
            return self._non_crossing(out_raw)

# ---------------------------------------------------------------
# Loss & LightningModule
# ---------------------------------------------------------------
def quantile_loss(y_pred, y_true, q_list):
    diff = y_true.unsqueeze(-1) - y_pred
    losses = [torch.max((q-1)*diff[..., i], q*diff[..., i]) for i, q in enumerate(q_list)]
    return torch.mean(torch.stack(losses, dim=-1))

class QuantileLightningModule(pl.LightningModule):
    def __init__(self, d_in, H, q_list, horz_dim, hidden, dropout, lr, weight_decay,
                 policy_ridge_lambda_base: float = 0.0,
                 policy_ridge_lambda_int: float = 0.0,   # (not used; kept for API stability)
                 policy_ridge_lambda_lag: float = 0.0,   # small L2 on lags is fine
                 anchor_ridge_lambda: float = 0.0,
                 season_ridge_lambda: float = 0.0,
                 enable_monotone_hidden: bool = True,
                 policy_feature_indices_neg=None,
                 policy_caps_vec=None):
        super().__init__()
        self.save_hyperparameters()
        self.model = HorizonQuantileModel(
            d_in=d_in, H=H, q_list=q_list,
            horz_dim=horz_dim, hidden=hidden, dropout=dropout,
            enable_monotone_hidden=enable_monotone_hidden,
            policy_feature_indices_neg=policy_feature_indices_neg,
            policy_caps_vec=policy_caps_vec,
        )
        self.q_list = list(q_list); self.lr = lr; self.weight_decay = weight_decay
        self.policy_ridge_lambda_base = float(policy_ridge_lambda_base)
        self.policy_ridge_lambda_lag  = float(policy_ridge_lambda_lag)
        self.anchor_ridge_lambda = float(anchor_ridge_lambda)
        self.season_ridge_lambda = float(season_ridge_lambda)
        self.d_in = d_in

    def _policy_ridge_penalty(self) -> torch.Tensor:
        W_in = self.model.fixed_head_weight_free[:, :self.d_in] if (getattr(self.model, "enable_monotone_hidden", False) and hasattr(self.model, "fixed_head_weight_free")) else self.model.fixed_head.weight[:, :self.d_in]
        names = KNOWN_COLS

        ewm_idx = [i for i, n in enumerate(names) if ("_ewm_hl" in n and n.startswith("covid_19_"))]
        lag_idx = [i for i, n in enumerate(names) if ("_lag" in n and n.startswith("covid_19_"))]
        anchor_idx = [i for i, n in enumerate(names) if n in ("prev_ma_4w_c","prev_slope_4w_c")]
        season_idx = [i for i, n in enumerate(names) if n in SEASONAL_FEATURES]

        pen = W_in.new_tensor(0.0)
        if ewm_idx:
            pen = pen + self.policy_ridge_lambda_base * (W_in[:, ewm_idx]**2).mean()
        if lag_idx:
            pen = pen + self.policy_ridge_lambda_lag  * (W_in[:, lag_idx]**2).mean()
        if anchor_idx:
            pen = pen + self.anchor_ridge_lambda * (W_in[:, anchor_idx]**2).mean()
        if season_idx:
            pen = pen + self.season_ridge_lambda * (W_in[:, season_idx]**2).mean()
        return pen

    def training_step(self, batch, batch_idx):
        X, y, _ = batch
        y_pred = self.model(X)
        loss_q = quantile_loss(y_pred, y, self.q_list)
        loss_pr = self._policy_ridge_penalty()
        loss = loss_q + loss_pr
        self.log("train_loss", loss, prog_bar=True, on_step=False, on_epoch=True)
        self.log("policy_ridge", loss_pr, prog_bar=False, on_step=False, on_epoch=True)
        return loss

    def validation_step(self, batch, batch_idx):
        X, y, _ = batch
        y_pred = self.model(X)
        loss_q = quantile_loss(y_pred, y, self.q_list)
        loss_pr = self._policy_ridge_penalty()
        loss = loss_q + loss_pr
        self.log("val_loss", loss, prog_bar=True, on_step=False, on_epoch=True)
        return loss

    def configure_optimizers(self):
        opt = torch.optim.Adam(self.parameters(), lr=self.lr, weight_decay=self.weight_decay)
        sch = torch.optim.lr_scheduler.ReduceLROnPlateau(opt, mode="min", patience=2, factor=0.6)
        return {"optimizer": opt, "lr_scheduler": {"scheduler": sch, "monitor": "val_loss"}}

# ---------------------------------------------------------------
# Hyperparameter Optimization (Optuna)
#   Only EWM family caps are tuned; lag families removed
# ---------------------------------------------------------------
LOG_DIR    = os.path.join(MODEL_DIR, "logs")
OPTUNA_DIR = os.path.join(MODEL_DIR, "optuna")
os.makedirs(LOG_DIR, exist_ok=True); os.makedirs(OPTUNA_DIR, exist_ok=True)

def objective(trial: optuna.Trial):
    horz_dim     = trial.suggest_int("horz_dim", 12, 12)
    hidden       = trial.suggest_int("hidden", 128, 128)
    dropout      = trial.suggest_float("dropout", 0.10, 0.20)
    lr           = trial.suggest_float("lr", 5e-4, 3e-3, log=True)
    weight_decay = trial.suggest_float("weight_decay", 3e-6, 3e-5, log=True)

    lambda_base   = trial.suggest_float("ridge_base",   1e-2, 1e-2, log=True)
    lambda_lag    = trial.suggest_float("ridge_lag",    1e-3, 1e-3, log=True)
    anchor_lambda = trial.suggest_float("ridge_anchor", 1e-2, 1e-2, log=True)
    season_lambda = trial.suggest_float("ridge_season", 1e-1, 1e-1, log=True)

    # EWM cap families only
    #cap_ewm6  = trial.suggest_float("cap_ewm6",  0.12, 0.24)
    #cap_ewm12 = trial.suggest_float("cap_ewm12", 0.16, 0.30)
    cap_ewm6  = trial.suggest_float("cap_ewm6",  0.12, 0.24)
    ratio     = trial.suggest_float("cap_ratio_12_over_6", 1.00, 1.35)
    cap_ewm12 = cap_ewm6 * ratio


    # rebuild per-name caps: ONLY for EWMs
    POLICY_CAPS_BY_NAME_TRIAL = {}
    for n in KNOWN_COLS:
        if n.endswith("_ewm_hl6"):    POLICY_CAPS_BY_NAME_TRIAL[n] = cap_ewm6
        elif n.endswith("_ewm_hl12"): POLICY_CAPS_BY_NAME_TRIAL[n] = cap_ewm12

    # indices & caps aligned to KNOWN_COLS (EWMs only)
    POLICY_NEG_IDX_TRIAL, POLICY_CAPS_VEC_TRIAL = [], []
    for i, n in enumerate(KNOWN_COLS):
        if n.startswith("covid_19_") and ("_ewm_hl" in n):
            POLICY_NEG_IDX_TRIAL.append(i)
            POLICY_CAPS_VEC_TRIAL.append(POLICY_CAPS_BY_NAME_TRIAL.get(n, 0.15))

    max_epochs = 50
    pl.seed_everything(SEED, workers=True)

    module = QuantileLightningModule(
        d_in=len(KNOWN_COLS), H=H, q_list=QUANTILES,
        horz_dim=horz_dim, hidden=hidden, dropout=dropout,
        lr=lr, weight_decay=weight_decay,
        policy_ridge_lambda_base=lambda_base,
        policy_ridge_lambda_lag=lambda_lag,   # keep small L2 on lags if desired
        anchor_ridge_lambda=anchor_lambda,
        season_ridge_lambda=season_lambda,
        enable_monotone_hidden=True,
        policy_feature_indices_neg=POLICY_NEG_IDX_TRIAL,
        policy_caps_vec=POLICY_CAPS_VEC_TRIAL,
    )

    ckpt_cb = ModelCheckpoint(monitor="val_loss", mode="min", save_top_k=1, save_weights_only=True)
    es_cb   = EarlyStopping(monitor="val_loss", mode="min", patience=3, min_delta=1e-4)
    prune_cb = PyTorchLightningPruningCallback(trial, monitor="val_loss")

    trainer = pl.Trainer(
        max_epochs=max_epochs,
        accelerator="gpu" if torch.cuda.is_available() else "cpu",
        devices=1,
        callbacks=[ckpt_cb, es_cb, prune_cb],
        log_every_n_steps=50,
        enable_checkpointing=True,
        precision="16-mixed" if torch.cuda.is_available() else "32-true",
        logger=False,
        deterministic=True,
        check_val_every_n_epoch=1,
        num_sanity_val_steps=0,
        gradient_clip_val=1.0
    )
    trainer.fit(module, train_dataloaders=train_loader, val_dataloaders=val_loader)
    trial.set_user_attr("best_ckpt_path", ckpt_cb.best_model_path)
    return float(ckpt_cb.best_model_score.item())

pruner = SuccessiveHalvingPruner(min_resource=15, reduction_factor=3, min_early_stopping_rate=0)
study = optuna.create_study(
    direction="minimize",
    sampler=optuna.samplers.TPESampler(seed=SEED),
    pruner=pruner
)
study.optimize(objective, n_trials=100, timeout=10800)

BEST_TRIAL_JSON = os.path.join(OPTUNA_DIR, "best_trial_omicron.json")
best_trial_payload = {
    "number": study.best_trial.number,
    "value": float(study.best_trial.value) if study.best_trial.value is not None else None,
    "params": {k: (float(v) if hasattr(v, "__float__") else v) for k, v in study.best_trial.params.items() },
}
with open(BEST_TRIAL_JSON, "w") as f: json.dump(best_trial_payload, f, indent=2)

# ---------------------------------------------------------------
# Build best module and (optionally) load best checkpoint
#   Rebuild cap vectors using best params (EWMs only)
# ---------------------------------------------------------------
bp = study.best_trial.params

# Recreate cap families from best params (fallback to defaults if absent)
cap_ewm6  = float(bp.get("cap_ewm6",  DEFAULT_CAPS_BY_SUFFIX["_ewm_hl6"]))
cap_ewm12 = float(bp.get("cap_ewm12", DEFAULT_CAPS_BY_SUFFIX["_ewm_hl12"]))

POLICY_CAPS_BY_NAME_BEST = {}
for n in KNOWN_COLS:
    if n.endswith("_ewm_hl6"):    POLICY_CAPS_BY_NAME_BEST[n] = cap_ewm6
    elif n.endswith("_ewm_hl12"): POLICY_CAPS_BY_NAME_BEST[n] = cap_ewm12

POLICY_NEG_IDX_BEST, POLICY_CAPS_VEC_BEST = [], []
for i, n in enumerate(KNOWN_COLS):
    if n.startswith("covid_19_") and ("_ewm_hl" in n):
        POLICY_NEG_IDX_BEST.append(i)
        POLICY_CAPS_VEC_BEST.append(POLICY_CAPS_BY_NAME_BEST.get(n, 0.15))

best_module = QuantileLightningModule(
    d_in=len(KNOWN_COLS),
    H=H,
    q_list=QUANTILES,
    horz_dim=int(bp.get("horz_dim", 8)),
    hidden=int(bp.get("hidden", 128)),
    dropout=float(bp.get("dropout", 0.0)),
    lr=float(bp.get("lr", 1e-3)),
    weight_decay=float(bp.get("weight_decay", 1e-5)),
    policy_ridge_lambda_base=float(bp.get("ridge_base", 0.0)),
    policy_ridge_lambda_lag =float(bp.get("ridge_lag",  0.0)),  # allowed
    anchor_ridge_lambda=float(bp.get("ridge_anchor", 0.0)),
    season_ridge_lambda=float(bp.get("ridge_season", 0.0)),
    enable_monotone_hidden=True,
    policy_feature_indices_neg=POLICY_NEG_IDX_BEST,
    policy_caps_vec=POLICY_CAPS_VEC_BEST,
)

_best_ckpt = study.best_trial.user_attrs.get("best_ckpt_path")
if _best_ckpt and os.path.isfile(_best_ckpt):
    state = torch.load(_best_ckpt, map_location="cpu")
    sd = state["state_dict"] if isinstance(state, dict) and "state_dict" in state else state
    best_module.load_state_dict(sd, strict=True)

# ---------------------------------------------------------------
# Export preprocessing artifacts (all with _omicron suffix)
# ---------------------------------------------------------------
SCHEMA_JSON   = os.path.join(MODEL_DIR, "serving_schema_omicron.json")
COUNTRY_MAP   = os.path.join(MODEL_DIR, "country_index_map_omicron.json")
VAL_COUNTRIES = os.path.join(MODEL_DIR, "val_countries_omicron.json")
CENTER_JSON   = os.path.join(MODEL_DIR, "center_means_omicron.json")

serving_notes = [
    "No PCA. Policies use RAW sliders.",
    "Policy windows are exponential weighted means (Koyck) with halflives {6,12}.",
    "Anchors: prev_ma_4w, prev_slope_4w (history-only); time_c fixed across horizon.",
    "Smoothed policy windows used directly (no residualization).",
    "Step-varying across horizon: policy windows & discrete lags; seasonality & latitude held fixed.",
    "Non-crossing quantiles via cumulative Softplus.",
    "Policy EWMs hard-constrained non-positive (per-feature caps). Lag features are unconstrained (optionally L2-regularized).",
    "Targets use stabilized logit on y/U; outputs are sigmoid-mapped logits and scaled by U at serve time.",
    "Single model; no random effects.",
    "Includes lag_gate feature; served with lag gate OFF (validation matches serving).",
]

serving_schema = {
    "time_idx": "week_id",
    "target": "covid_19_prevalence",
    "group_ids": ["country_iso3"],
    "max_prediction_length": int(H),
    "static_categoricals": ["country_iso3"],
    "static_reals": [],
    "time_varying_known_reals": list(KNOWN_COLS),
    "constants": {
        "prevalence_cap": float(TARGET_CAP_U),
    },
    "time_varying_unknown_reals": [],
    "normalization": {"type": "FeatureZNormOnly"},
    "categorical_encoders": {"country_iso3": {"encoder": "LabelMap", "add_nan": False}},
    "notes": serving_notes,
}
with open(SCHEMA_JSON, "w") as f: json.dump(serving_schema, f, indent=2)

with open(COUNTRY_MAP, "w") as f: json.dump(country_index_map, f, indent=2)
with open(VAL_COUNTRIES, "w") as f: json.dump(sorted(list(val_countries)), f, indent=2)
with open(CENTER_JSON, "w") as f: json.dump(CENTER_MEANS_FOR_INTS, f, indent=2)

# ---------------------------------------------------------------
# TorchScript export (PF-free serving)
# ---------------------------------------------------------------
SCRIPTED_PT      = os.path.join(MODEL_DIR, "model_scripted_omicron.pt")
FEATURE_CONTRACT = os.path.join(MODEL_DIR, "feature_contract_omicron.json")

best_module.eval().to("cpu")
model_for_export = best_module.model.eval().cpu()

if len(val_dataset) == 0:
    raise RuntimeError(
        "Validation set is empty after horizon construction. "
        "Try reducing H (e.g., H=8) or adjusting the train/val split."
    )

ex_X, _, _ = val_dataset[0]
ex_X = ex_X.unsqueeze(0)

class _ModelScriptWrapper(torch.nn.Module):
    def __init__(self, core_model: nn.Module, cap_u: float):
        super().__init__(); self.core = core_model; self.cap_u = float(cap_u)
    def forward(self, decoder_cont: torch.Tensor) -> torch.Tensor:
        return self.cap_u * torch.sigmoid(self.core(decoder_cont))

with torch.inference_mode():
    scripted = torch.jit.trace(_ModelScriptWrapper(model_for_export, cap_u=TARGET_CAP_U).eval(),
                               (ex_X.float(),),
                               strict=False)
scripted.save(SCRIPTED_PT)

contract = {
    "decoder_cont_lastdim": int(ex_X.shape[-1]),
    "expects": {"decoder_cont": "float32 [B, H, D_known]"},
    "produces": f"float32 [B, H, Q] in [0,{TARGET_CAP_U}] (sigmoid on logits, scaled by U)",
    "horizon": int(H),
    "quantiles": [float(q) for q in QUANTILES],
    "feature_order": list(KNOWN_COLS),
}
with open(FEATURE_CONTRACT, "w") as f: json.dump(contract, f, indent=2)

print("✅ TorchScript saved to:", SCRIPTED_PT)
print("📝 Feature contract saved to:", FEATURE_CONTRACT)

# ---------------------------------------------------------------
# Bounded-output smoke test
# ---------------------------------------------------------------
with torch.no_grad():
    logits = best_module.model(ex_X.float())     # [1,H,Q] logits
    scaled = TARGET_CAP_U * torch.sigmoid(logits)
    m = float(torch.max(scaled))
    assert m <= TARGET_CAP_U + 1e-6, f"Scaled outputs exceed U: {m} > {TARGET_CAP_U}"
print(f"✅ Bounded-output check passed (max ≤ U={TARGET_CAP_U:.6f})")

# ---------------------------------------------------------------
# Sanity checks
# ---------------------------------------------------------------
assert list(KNOWN_COLS) == serving_schema["time_varying_known_reals"], "KNOWN_COLS mismatch serve schema."

# ---------------------------------------------------------------
# Summary
# ---------------------------------------------------------------
def _mb(path): return os.path.getsize(path)/(1024*1024) if os.path.exists(path) else 0.0
print("\n✅ Export complete →", MODEL_DIR)
for p in [
    os.path.join(MODEL_DIR, "model_scripted_omicron.pt"),
    os.path.join(MODEL_DIR, "feature_contract_omicron.json"),
    os.path.join(MODEL_DIR, "serving_schema_omicron.json"),
    os.path.join(MODEL_DIR, "country_index_map_omicron.json"),
    os.path.join(MODEL_DIR, "val_countries_omicron.json"),
    os.path.join(MODEL_DIR, "feature_norm_stats_omicron.json"),
    os.path.join(MODEL_DIR, "center_means_omicron.json"),
    os.path.join(MODEL_DIR, "best_trial_omicron.json"),
]:
    if os.path.exists(p):
        print(f" - {os.path.basename(p):28s}: {_mb(p):.2f} MB")
```

```python
# ================================================================
# TRAIN-BEST PIPELINE (POST-HPO)
# Horizon-Aware Quantile Regressor — Train final model with fixed best params
#   - Uses best params saved by HPO (best_trial_omicron.json)
#   - Trains up to 250 epochs with EarlyStopping(patience=10)
#   - EWMs-only negative caps (lags unconstrained; optional L2 on lags)
#   - Validation mirrors serving: lag gate OFF
#   - Exports scripted model + schema/contract artifacts with _omicron suffix
# ================================================================

# If running in Colab, uncomment the next line:
!pip install -q "lightning==2.4.*"

import os, json, random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

import lightning.pytorch as pl
from lightning.pytorch.callbacks import EarlyStopping, ModelCheckpoint

from google.colab import drive

# ---------------------------------------------------------------
# Repro & setup
# ---------------------------------------------------------------
SEED = 42
pl.seed_everything(SEED, workers=True)
np.random.seed(SEED); random.seed(SEED); torch.manual_seed(SEED)
torch.use_deterministic_algorithms(True, warn_only=True)

drive.mount("/content/drive")

# Paths
DATA_PATH = "/content/drive/MyDrive/Data/df_final_omicron.csv"
MODEL_DIR = "/content/drive/MyDrive/Models"
OPTUNA_DIR = os.path.join(MODEL_DIR, "optuna")  # where best_trial_omicron.json lives
os.makedirs(MODEL_DIR, exist_ok=True); os.makedirs(OPTUNA_DIR, exist_ok=True)

BEST_TRIAL_JSON = os.path.join(OPTUNA_DIR, "best_trial_omicron.json")

# ---------------------------------------------------------------
# Load data
# ---------------------------------------------------------------
df = pd.read_csv(DATA_PATH)

# time alias used for centering
df["time"] = df["week_id"]

required_cols = {
    "country_iso3","year","week_id","iso_week","time",
    "covid_19_prevalence",
    "covid_19_policy_stringency","covid_19_face_covering_policy","covid_19_testing_tracing_policy",
    "latitude","sine_seasonality","cosine_seasonality",
}
missing = sorted(list(required_cols - set(df.columns)))
if missing:
    raise ValueError(f"Missing required columns: {missing}")

df["week_id"] = pd.to_numeric(df["week_id"], errors="raise")
df = df.sort_values(["country_iso3","week_id"]).reset_index(drop=True)

print("shape:", df.shape)
print("#countries:", df["country_iso3"].nunique())
print("years:", df["year"].min(), "→", df["year"].max())

# ---------------------------------------------------------------
# Train/val split by country
# ---------------------------------------------------------------
unique_countries = df["country_iso3"].unique().tolist()
random.shuffle(unique_countries)
k_val = max(1, int(0.2 * len(unique_countries)))
val_countries = set(unique_countries[:k_val])

df_train = df[~df["country_iso3"].isin(val_countries)].copy()
df_val   = df[df["country_iso3"].isin(val_countries)].copy()

# ---------------------------------------------------------------
# Policy sliders (raw; no demeaning)
# ---------------------------------------------------------------
POLICY_COLS = (
    "covid_19_policy_stringency",
    "covid_19_face_covering_policy",
    "covid_19_testing_tracing_policy",
)

# ---------------------------------------------------------------
# Koyck smoothing windows (built on RAW policy columns)
# ---------------------------------------------------------------
HALFLIVES = (6, 12)
EWM_TAGS  = [f"ewm_hl{hl}" for hl in HALFLIVES]

def build_policy_koyck(df_in: pd.DataFrame, group_col: str = "country_iso3") -> pd.DataFrame:
    df = df_in.copy()
    for hl in HALFLIVES:
        tag = f"ewm_hl{hl}"
        for feat in POLICY_COLS:
            sm = (
                df.groupby(group_col)[feat]
                  .apply(lambda s: s.ewm(halflife=hl, adjust=False, min_periods=1).mean())
                  .reset_index(level=0, drop=True)
            )
            df[f"{feat}_{tag}"] = sm.astype(float)
    return df

# ------------------ Explicit policy lags ------------------
LAG_STEPS = (1, 2, 3, 4)

def add_policy_lags(
    df_in: pd.DataFrame,
    policy_cols=POLICY_COLS,
    group_col: str = "country_iso3",
    lag_steps=LAG_STEPS
) -> pd.DataFrame:
    """Add simple discrete lags of raw policy sliders (per-country)."""
    df = df_in.sort_values([group_col, "week_id"]).copy()
    for feat in policy_cols:
        g = df.groupby(group_col)[feat]
        for k in lag_steps:
            col = f"{feat}_lag{k}"
            df[col] = g.shift(k).fillna(df[feat]).astype(float)
    return df

# ---------------------------------------------------------------
# Anchor features (history-only)
# ---------------------------------------------------------------
def add_anchor_history_features(
    df: pd.DataFrame,
    country_col: str = "country_iso3",
    time_col: str = "week_id",
    prevalence_col: str = "covid_19_prevalence",
    trend_window: int = 4,
) -> pd.DataFrame:
    df = df.sort_values([country_col, time_col]).copy()
    df[prevalence_col] = pd.to_numeric(df[prevalence_col], errors="coerce").fillna(0.0)

    # rolling mean (anchor)
    df["prev_ma_4w"] = (
        df.groupby(country_col)[prevalence_col]
          .rolling(window=trend_window, min_periods=1).mean()
          .reset_index(level=0, drop=True).astype(float)
    )

    # rolling slope (anchor)
    def _rolling_slope(series: pd.Series) -> float:
        y = series.to_numpy(dtype=float); k = y.shape[0]
        if k < 2: return 0.0
        x = np.arange(k, dtype=float); xm, ym = x.mean(), y.mean()
        denom = ((x - xm)**2).sum()
        if denom <= 0: return 0.0
        return float(((x - xm)*(y - ym)).sum() / denom)

    df["prev_slope_4w"] = (
        df.groupby(country_col)[prevalence_col]
          .rolling(window=trend_window, min_periods=2).apply(_rolling_slope, raw=False)
          .reset_index(level=0, drop=True).fillna(0.0).astype(float)
    )
    return df

# ---------------------------------------------------------------
# Synthetic countries (optional; default: enabled)
# ---------------------------------------------------------------
USE_SYNTH = True
SYNTH_BASE_BLEND_FEATURES = [
    "covid_19_prevalence",
    "covid_19_policy_stringency","covid_19_face_covering_policy","covid_19_testing_tracing_policy",
    "latitude",
]
SYNTH_PASSTHROUGH = ["week_id","year","iso_week","time","sine_seasonality","cosine_seasonality"]

def synthesize_countries(
    df_in: pd.DataFrame,
    country_col: str = "country_iso3",
    time_col: str = "week_id",
    ratios: tuple = (0.3, 0.5, 0.7),
    max_pairs: int = 26,
    same_hemisphere_only: bool = True,
    equator_band: float = 0.0,
    sample_with_replacement: bool = True,
    center_means: dict | None = None,
):
    countries = df_in[country_col].astype(str).unique().tolist()
    country_lat = (
        df_in[[country_col, "latitude"]]
        .dropna(subset=["latitude"]).drop_duplicates(subset=[country_col])
        .set_index(country_col)["latitude"]
    )
    countries = [c for c in countries if c in country_lat.index]
    def hemi_code(lat: float) -> int:
        lat = float(lat)
        return 0 if abs(lat) <= float(equator_band) else (1 if lat > 0.0 else -1)
    hemi = country_lat.apply(hemi_code)

    from itertools import combinations
    all_pairs = list(combinations(countries, 2))
    valid_pairs = [(c1,c2) for (c1,c2) in all_pairs if (hemi[c1] == hemi[c2])] if same_hemisphere_only else all_pairs
    if not valid_pairs:
        return pd.DataFrame(columns=[time_col, country_col] + SYNTH_BASE_BLEND_FEATURES + SYNTH_PASSTHROUGH)
    random.shuffle(valid_pairs)
    pairs = (valid_pairs[:max_pairs] if len(valid_pairs) >= max_pairs
             else (random.choices(valid_pairs, k=max_pairs) if sample_with_replacement else valid_pairs))

    numeric_cols = df_in.columns[df_in.dtypes.apply(lambda d: np.issubdtype(d, np.number))].tolist()
    frames = []
    for c1, c2 in pairs:
        s1 = df_in[df_in[country_col]==c1].set_index(time_col)
        s2 = df_in[df_in[country_col]==c2].set_index(time_col)
        idx = s1.index.intersection(s2.index)
        if len(idx)==0: continue
        s1, s2 = s1.loc[idx], s2.loc[idx]
        for r in ratios:
            out = pd.DataFrame(index=idx)
            for col in SYNTH_BASE_BLEND_FEATURES:
                if col in numeric_cols: out[col] = (1-r)*s1[col].astype(float) + r*s2[col].astype(float)
            for pc in SYNTH_PASSTHROUGH:
                if pc in s1.columns: out[pc] = s1[pc].values
            out = out.reset_index(); out[country_col] = f"{c1}_{c2}_mix_{int(r*100)}"
            frames.append(out)

    if not frames:
        return pd.DataFrame(columns=[time_col, country_col] + SYNTH_BASE_BLEND_FEATURES + SYNTH_PASSTHROUGH)

    df_blend = (pd.concat(frames, ignore_index=True)
                .sort_values([country_col, time_col]).reset_index(drop=True))
    if time_col in df_blend and pd.api.types.is_float_dtype(df_blend[time_col]): df_blend[time_col] = df_blend[time_col].round().astype(int)
    if "year" in df_blend and pd.api.types.is_float_dtype(df_blend["year"]): df_blend["year"] = df_blend["year"].round().astype(int)

    # Build windows + anchors BEFORE centering
    df_blend = build_policy_koyck(df_blend, group_col=country_col)
    df_blend = add_policy_lags(df_blend, group_col=country_col)
    df_blend = add_anchor_history_features(df_blend, country_col=country_col, time_col=time_col,
                                           prevalence_col="covid_19_prevalence", trend_window=4)

    # Center moderators (TRAIN means) — only after anchors exist
    CENTER_KEYS = ["prev_ma_4w","prev_slope_4w","time"]
    for col in CENTER_KEYS:
        if col not in df_blend.columns: continue
        mu = float(center_means[col]) if (center_means is not None and col in center_means) else float(df_blend[col].mean())
        df_blend[f"{col}_c"] = df_blend[col].astype(float) - mu

    return df_blend

# ---------------------------------------------------------------
# Build train/val features
# ---------------------------------------------------------------
df_train_feat = build_policy_koyck(df_train, group_col="country_iso3")
df_val_feat   = build_policy_koyck(df_val,   group_col="country_iso3")

df_train_feat = add_policy_lags(df_train_feat, group_col="country_iso3")
df_val_feat   = add_policy_lags(df_val_feat,   group_col="country_iso3")

df_train_feat = add_anchor_history_features(df_train_feat, country_col="country_iso3", time_col="week_id",
                                            prevalence_col="covid_19_prevalence", trend_window=4)
df_val_feat   = add_anchor_history_features(df_val_feat,   country_col="country_iso3", time_col="week_id",
                                            prevalence_col="covid_19_prevalence", trend_window=4)

# Center moderators AFTER anchors exist; only what's used
CENTER_FOR_INTS = ["prev_ma_4w","prev_slope_4w","time"]
CENTER_MEANS_FOR_INTS = df_train_feat[CENTER_FOR_INTS].astype(float).mean().to_dict()
for col in CENTER_FOR_INTS:
    mu = float(CENTER_MEANS_FOR_INTS[col])
    df_train_feat[f"{col}_c"] = df_train_feat[col].astype(float) - mu
    df_val_feat[f"{col}_c"]   = df_val_feat[col].astype(float)   - mu

# Interactions are currently OFF (no *_relu moderators in use)

# Seasonality × latitude interactions (used)
def build_interactions(df_in: pd.DataFrame) -> pd.DataFrame:
    df = df_in.copy()
    df["sine_seasonality_x_latitude"]   = df["sine_seasonality"].astype(float) * df["latitude"].astype(float)
    df["cosine_seasonality_x_latitude"] = df["cosine_seasonality"].astype(float) * df["latitude"].astype(float)
    return df

df_train_feat = build_interactions(df_train_feat)
df_val_feat   = build_interactions(df_val_feat)

# Optional synthetic augmentation
if USE_SYNTH:
    df_synth = synthesize_countries(
        df_in=df_train,
        country_col="country_iso3",
        time_col="week_id",
        center_means=CENTER_MEANS_FOR_INTS
    )
    if len(df_synth):
        df_synth = build_interactions(df_synth)
        df_synth = df_synth[df_synth["year"] > 2019].copy()
        df_train_feat = pd.concat([df_train_feat, df_synth], ignore_index=True)

# Focus on pandemic years
df_train_feat = df_train_feat[df_train_feat["year"] > 2019].copy()
df_val_feat   = df_val_feat[df_val_feat["year"] > 2019].copy()

# ---------------------------------------------------------------
# Horizon supervision
# ---------------------------------------------------------------
H = 12

def _policy_base_names():
    base = [f"{feat}_{tag}" for tag in EWM_TAGS for feat in POLICY_COLS]
    return base, []  # (policy EWMs, interactions OFF)

policy_base, policy_allmods = _policy_base_names()

# Lag feature basenames
POLICY_LAG_BASENAMES = [f"{feat}_lag{k}" for feat in POLICY_COLS for k in LAG_STEPS]

# handy name list for lag columns
LAG_FEATURES = [f"{feat}_lag{k}" for feat in POLICY_COLS for k in LAG_STEPS]

# KNOWN_COLS … append lag_gate at the end
KNOWN_COLS = [
    # anchors / time (centered where *_c is explicit)
    "prev_ma_4w_c","prev_slope_4w_c","time_c",
    # geography & seasonality
    "latitude","sine_seasonality","cosine_seasonality",
    "sine_seasonality_x_latitude","cosine_seasonality_x_latitude",
    # policies: discrete lags + Koyck windows
    *POLICY_LAG_BASENAMES,     # lags
    *policy_base, *policy_allmods,
    # (NEW)
    "lag_gate",
]

def build_horizon_supervision(
    df_in: pd.DataFrame, country_col="country_iso3", time_col="week_id",
    target_col="covid_19_prevalence", H=8, c2i: dict[str,int] | None = None,
):
    df = df_in.copy()
    if c2i is None:
        countries = sorted(df[country_col].astype(str).unique().tolist())
        c2i = {c: i for i, c in enumerate(countries)}
    df["country_idx"] = df[country_col].astype(str).map(c2i).astype(int)
    for h in range(1, H+1):
        df[f"y_h{h}"] = df.groupby(country_col)[target_col].shift(-h)
    keep = np.ones(len(df), dtype=bool)
    for h in range(1, H+1):
        keep &= df[f"y_h{h}"].notna().values
    return df.loc[keep].reset_index(drop=True), c2i

all_countries = sorted(set(df_train_feat["country_iso3"].astype(str)) | set(df_val_feat["country_iso3"].astype(str)))
country_index_map = {c: i for i, c in enumerate(all_countries)}

df_train_sup, _ = build_horizon_supervision(df_train_feat, H=H, c2i=country_index_map)
df_val_sup,   _ = build_horizon_supervision(df_val_feat,   H=H, c2i=country_index_map)

# ---------------------------------------------------------------
# Feature normalization (train-only)
# ---------------------------------------------------------------
FEATURE_NORM_JSON = os.path.join(MODEL_DIR, "feature_norm_stats_omicron.json")

for col in KNOWN_COLS:
    if col not in df_train_feat.columns: df_train_feat[col] = 0.0
    if col not in df_val_feat.columns:   df_val_feat[col]   = 0.0

_train_X = df_train_feat[KNOWN_COLS].astype(float)
train_means = _train_X.mean(axis=0)
train_stds  = _train_X.std(axis=0).replace(0.0, 1.0)

df_train_feat[KNOWN_COLS] = (df_train_feat[KNOWN_COLS].astype(float) - train_means) / train_stds
df_val_feat[KNOWN_COLS]   = (df_val_feat[KNOWN_COLS].astype(float)   - train_means) / train_stds

feature_norm = {col: {"mean": float(train_means[col]), "std": float(train_stds[col])} for col in KNOWN_COLS}

with open(FEATURE_NORM_JSON, "w") as f: json.dump(feature_norm, f, indent=2)
print(f"✅ Feature normalization written → {FEATURE_NORM_JSON}")

df_train_sup, _ = build_horizon_supervision(df_train_feat, H=H, c2i=country_index_map)
df_val_sup,   _ = build_horizon_supervision(df_val_feat,   H=H, c2i=country_index_map)

# ---------------------------------------------------------------
# Targets: stabilized logit on y/U with cap scaling
# ---------------------------------------------------------------
TARGET_CAP_U = 1.00
EPS_LOGIT = 1e-6

def _logit01(y: np.ndarray, eps: float = EPS_LOGIT) -> np.ndarray:
    y = np.asarray(y, dtype=np.float64)
    y = np.clip(y, eps, 1.0 - eps)
    return np.log(y) - np.log1p(-y)

def make_logit_targets_scaled(df_in: pd.DataFrame, H: int, U: float, target_prefix_out: str = "y_h"):
    df = df_in.copy()
    invU = 1.0 / float(U)
    for h in range(1, H+1):
        col = f"{target_prefix_out}{h}"
        y = df[col].to_numpy(dtype=float) * invU
        z = _logit01(y, eps=EPS_LOGIT)
        df[f"{col}_logit"] = z.astype(np.float32)
    return df

df_train_sup = make_logit_targets_scaled(df_train_sup, H=H, U=TARGET_CAP_U)
df_val_sup   = make_logit_targets_scaled(df_val_sup,   H=H, U=TARGET_CAP_U)

with open(FEATURE_NORM_JSON, "r") as _f:
    NORM_STATS = json.load(_f)

# ---------------------------------------------------------------
# Torch datasets & loaders
# ---------------------------------------------------------------
from torch.utils.data import Dataset, DataLoader
QUANTILES = [0.1, 0.5, 0.9]

# STEP_VARY_COLS: features that update over horizon
SEASONAL_FEATURES = [
    "sine_seasonality","cosine_seasonality",
    "sine_seasonality_x_latitude","cosine_seasonality_x_latitude",
]
STEP_VARY_COLS = set([
    *[f"{feat}_{tag}" for tag in EWM_TAGS for feat in POLICY_COLS],  # Koyck windows
    *[f"{feat}_lag{k}" for feat in POLICY_COLS for k in LAG_STEPS],  # discrete lags
    *SEASONAL_FEATURES,
])

# (interactions are OFF; function kept for compatibility if later enabled)
INT_MODS: list[str] = []

def _recompute_policy_interactions_inplace(xh: dict, x_anchor: dict, norm_stats: dict):
    # No-op when INT_MODS is empty; kept for drop-in compatibility.
    return

class HorizonDataset(Dataset):
    def __init__(self, df_in, known_cols, H, country_col="country_iso3", time_col="week_id",
                 norm_stats: dict | None = None,
                 gate_mode: str = "fixed",   # "sample" or "fixed"
                 gate_p: float = 0.5,        # only used if gate_mode="sample"
                 gate_value: int = 1):       # only used if gate_mode="fixed"
        self.df = df_in.reset_index(drop=True)
        self.known_cols = known_cols
        self.H = H
        self.country_col = country_col
        self.time_col = time_col
        self.norm_stats = norm_stats or {}
        self.gate_mode  = gate_mode
        self.gate_p     = float(gate_p)
        self.gate_value = int(gate_value)

        cols_needed = list(set(known_cols) | {country_col, time_col, *INT_MODS})
        slim = self.df[cols_needed].copy()
        self._by_key = {(str(r[country_col]), int(r[time_col])): r for _, r in slim.iterrows()}

        # cache indices of lag features in KNOWN_COLS for fast gating
        self.lag_idx = [i for i, n in enumerate(self.known_cols) if n in LAG_FEATURES]
        self.gate_idx = self.known_cols.index("lag_gate")

    def __len__(self):
        return len(self.df)

    def _sample_gate(self):
        if self.gate_mode == "sample":
            return 1.0 if (np.random.rand() < self.gate_p) else 0.0
        return float(self.gate_value)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        ctry = str(row[self.country_col])
        t0   = int(row[self.time_col])

        # base (already Z-scored in the dataframe)
        x0 = {c: float(row[c]) if c in row and pd.notna(row[c]) else 0.0 for c in self.known_cols}

        # choose the gate for this sample
        gate = self._sample_gate()
        x0["lag_gate"] = gate  # in Z-space; 0/1 fine after Z-norm

        # Anchor values for interactions (inactive in your config, kept for completeness)
        x_anchor = {m: (float(row[m]) if (m in row and pd.notna(row[m])) else 0.0) for m in INT_MODS}
        x_anchor.update({c: x0.get(c, 0.0) for c in self.known_cols})

        Xh = np.zeros((self.H, len(self.known_cols)), dtype=np.float32)
        for h in range(1, self.H+1):
            xh = dict(x0)
            fut = self._by_key.get((ctry, t0 + h), None)
            if fut is not None:
                for col in STEP_VARY_COLS:
                    if col in self.known_cols and col in fut and pd.notna(fut[col]):
                        xh[col] = float(fut[col])

            _recompute_policy_interactions_inplace(xh, x_anchor, self.norm_stats)

            # apply the gate to all lag features (multiply in Z-space)
            if self.lag_idx:
                for i in self.lag_idx:
                    n = self.known_cols[i]
                    xh[n] = float(xh.get(n, 0.0)) * gate

            # write row in KNOWN_COLS order
            Xh[h-1, :] = np.fromiter((xh.get(c, 0.0) for c in self.known_cols),
                                     dtype=np.float32, count=len(self.known_cols))

        y = np.stack([row[f"y_h{h}_logit"] for h in range(1, self.H+1)], axis=0).astype(np.float32)
        country_idx = int(row["country_idx"])
        return torch.from_numpy(Xh), torch.from_numpy(y), torch.tensor(country_idx, dtype=torch.long)

# Create loaders with gating settings
train_dataset = HorizonDataset(
    df_train_sup, KNOWN_COLS, H, norm_stats=NORM_STATS,
    gate_mode="sample", gate_p=0.5   # mix of on/off during training
)

# ✅ Validation mirrors serving (lag gate OFF)
val_dataset   = HorizonDataset(
    df_val_sup, KNOWN_COLS, H, norm_stats=NORM_STATS,
    gate_mode="fixed", gate_value=0
)

batch_size = 1024
train_loader = DataLoader(train_dataset,
                          batch_size=batch_size,
                          shuffle=True,
                          num_workers=4,
                          pin_memory=True,
                          persistent_workers=True,
                          prefetch_factor=2,
                          drop_last=False)
val_loader   = DataLoader(val_dataset,
                          batch_size=batch_size*2,
                          shuffle=False,
                          num_workers=4,
                          pin_memory=True,
                          persistent_workers=True,
                          prefetch_factor=2,
                          drop_last=False)

# ---------------------------------------------------------------
# Per-feature negative caps for policy EWMs ONLY (lags unconstrained)
# ---------------------------------------------------------------
DEFAULT_CAPS_BY_SUFFIX = {
    "_ewm_hl6":  0.10,
    "_ewm_hl12": 0.12,
}

POLICY_CAPS_BY_NAME = {}
for n in KNOWN_COLS:
    if n.startswith("covid_19_") and ("_ewm_hl" in n):
        for suf, cap in DEFAULT_CAPS_BY_SUFFIX.items():
            if n.endswith(suf):
                POLICY_CAPS_BY_NAME[n] = cap
                break

POLICY_NEG_IDX = []
POLICY_CAPS_VEC = []
for i, n in enumerate(KNOWN_COLS):
    is_policy = n.startswith("covid_19_")
    is_ewm    = ("_ewm_hl" in n)
    if is_policy and is_ewm:
        POLICY_NEG_IDX.append(i)
        cap = (
            POLICY_CAPS_BY_NAME.get(n, None) if n in POLICY_CAPS_BY_NAME else
            (0.10 if n.endswith("_ewm_hl6")  else
             0.12 if n.endswith("_ewm_hl12") else 0.15)
        )
        POLICY_CAPS_VEC.append(cap)

# ---------------------------------------------------------------
# Seasonality (mains + their interaction with latitude)
# ---------------------------------------------------------------
SEASONAL_FEATURES = [
    "sine_seasonality","cosine_seasonality",
    "sine_seasonality_x_latitude","cosine_seasonality_x_latitude",
]

# ---------------------------------------------------------------
# Model (single model)
# ---------------------------------------------------------------
class HorizonQuantileModel(nn.Module):
    def __init__(self,
                 d_in: int,
                 H: int,
                 q_list=(0.1, 0.5, 0.9),
                 horz_dim: int = 8,
                 hidden: int | None = 128,
                 dropout: float = 0.0,
                 enable_monotone_hidden: bool = True,
                 # vectorized caps for selected policy features
                 policy_feature_indices_neg=None,
                 policy_caps_vec=None,
                 # backward-compat aliases (optional)
                 policy_idx_neg: torch.Tensor | None = None,
                 policy_caps_neg: torch.Tensor | None = None):
        super().__init__()
        self.d_in = d_in
        self.H = H
        self.q_list = list(q_list)
        self.Q = len(self.q_list)

        self.h_emb = nn.Embedding(H, horz_dim)

        self.fe_in = d_in + horz_dim
        self.use_hidden = hidden is not None
        self.dropout = nn.Dropout(dropout) if (self.use_hidden and dropout > 0) else nn.Identity()

        if not self.use_hidden:
            self.fixed_head  = nn.Linear(self.fe_in, 1, bias=False)
            self.quantile_out = nn.Linear(1, self.Q, bias=True)
        else:
            self.fixed_head  = nn.Linear(self.fe_in, hidden, bias=True)
            self.quantile_out = nn.Linear(hidden, self.Q, bias=True)

        # Monotone path (sign clamps on specified policy features)
        self.enable_monotone_hidden = bool(enable_monotone_hidden and self.use_hidden)

        if self.enable_monotone_hidden:
            W0 = self.fixed_head.weight.detach().clone()
            self.fixed_head_weight_free = nn.Parameter(W0)
            self.fixed_head_bias_param = self.fixed_head.bias

            neg = policy_feature_indices_neg if policy_feature_indices_neg is not None else policy_idx_neg
            caps = policy_caps_vec if policy_caps_vec is not None else policy_caps_neg
            neg = neg if neg is not None else []
            caps = caps if caps is not None else [0.15] * (len(neg) if hasattr(neg, "__len__") else 0)

            if isinstance(neg, torch.Tensor):
                neg_t = neg.to(dtype=torch.long)
            else:
                neg_t = torch.tensor(neg, dtype=torch.long)
            if isinstance(caps, torch.Tensor):
                caps_t = caps.to(dtype=torch.float32)
            else:
                caps_t = torch.tensor(caps, dtype=torch.float32)

            self.register_buffer("_policy_idx_neg", neg_t)
            self.register_buffer("_policy_caps", caps_t)

            qW0 = self.quantile_out.weight.detach().clone()
            self.quantile_out_weight_free = nn.Parameter(qW0)
            self.quantile_out_bias_param = self.quantile_out.bias

    @staticmethod
    def _non_crossing(q_raw: torch.Tensor) -> torch.Tensor:
        q10 = q_raw[..., 0:1]
        d2  = F.softplus(q_raw[..., 1:2])
        d3  = F.softplus(q_raw[..., 2:3])
        q50 = q10 + d2
        q90 = q50 + d3
        return torch.cat([q10, q50, q90], dim=-1)

    def forward(self, decoder_cont: torch.Tensor) -> torch.Tensor:
        B, H, D = decoder_cont.shape
        assert H == self.H and D == self.d_in

        h_ids = torch.arange(self.H, device=decoder_cont.device).view(1, H).expand(B, H)
        h_vec = self.h_emb(h_ids)

        fe = torch.cat([decoder_cont, h_vec], dim=-1)  # [B,H,fe_in]

        if self.enable_monotone_hidden:
            W_free = self.fixed_head_weight_free
            eff_W = W_free.clone()

            # per-index caps for all selected policy features (EWMs only)
            if self._policy_idx_neg.numel() > 0:
                sel  = eff_W[:, self._policy_idx_neg]             # [hidden, K]
                caps = self._policy_caps.view(1, -1)              # [1, K]
                eff_W[:, self._policy_idx_neg] = -(torch.sigmoid(sel) * caps)

            h1 = fe @ eff_W.t()
            if self.fixed_head_bias_param is not None:
                h1 = h1 + self.fixed_head_bias_param.view(1, 1, -1)
            h1 = F.relu(h1)
            h1 = self.dropout(h1)

            qW = F.softplus(self.quantile_out_weight_free)
            out_raw = h1 @ qW.t()
            if self.quantile_out_bias_param is not None:
                out_raw = out_raw + self.quantile_out_bias_param.view(1, 1, -1)
            return self._non_crossing(out_raw)

        if not self.use_hidden:
            mu = self.fixed_head(fe)
            out_raw = self.quantile_out(mu)
            return self._non_crossing(out_raw)
        else:
            h1 = self.fixed_head(fe)
            h1 = F.relu(h1); h1 = self.dropout(h1)
            out_raw = self.quantile_out(h1)
            return self._non_crossing(out_raw)

# ---------------------------------------------------------------
# Loss & LightningModule
# ---------------------------------------------------------------
def quantile_loss(y_pred, y_true, q_list):
    diff = y_true.unsqueeze(-1) - y_pred
    losses = [torch.max((q-1)*diff[..., i], q*diff[..., i]) for i, q in enumerate(q_list)]
    return torch.mean(torch.stack(losses, dim=-1))

class QuantileLightningModule(pl.LightningModule):
    def __init__(self, d_in, H, q_list, horz_dim, hidden, dropout, lr, weight_decay,
                 policy_ridge_lambda_base: float = 0.0,
                 policy_ridge_lambda_int: float = 0.0,   # (not used; kept for API stability)
                 policy_ridge_lambda_lag: float = 0.0,   # small L2 on lags is fine
                 anchor_ridge_lambda: float = 0.0,
                 season_ridge_lambda: float = 0.0,
                 enable_monotone_hidden: bool = True,
                 policy_feature_indices_neg=None,
                 policy_caps_vec=None):
        super().__init__()
        self.save_hyperparameters()
        self.model = HorizonQuantileModel(
            d_in=d_in, H=H, q_list=q_list,
            horz_dim=horz_dim, hidden=hidden, dropout=dropout,
            enable_monotone_hidden=enable_monotone_hidden,
            policy_feature_indices_neg=policy_feature_indices_neg,
            policy_caps_vec=policy_caps_vec,
        )
        self.q_list = list(q_list); self.lr = lr; self.weight_decay = weight_decay
        self.policy_ridge_lambda_base = float(policy_ridge_lambda_base)
        self.policy_ridge_lambda_lag  = float(policy_ridge_lambda_lag)
        self.anchor_ridge_lambda = float(anchor_ridge_lambda)
        self.season_ridge_lambda = float(season_ridge_lambda)
        self.d_in = d_in

    def _policy_ridge_penalty(self) -> torch.Tensor:
        W_in = self.model.fixed_head_weight_free[:, :self.d_in] if (getattr(self.model, "enable_monotone_hidden", False) and hasattr(self.model, "fixed_head_weight_free")) else self.model.fixed_head.weight[:, :self.d_in]
        names = KNOWN_COLS

        ewm_idx = [i for i, n in enumerate(names) if ("_ewm_hl" in n and n.startswith("covid_19_"))]
        lag_idx = [i for i, n in enumerate(names) if ("_lag" in n and n.startswith("covid_19_"))]
        anchor_idx = [i for i, n in enumerate(names) if n in ("prev_ma_4w_c","prev_slope_4w_c")]
        season_idx = [i for i, n in enumerate(names) if n in SEASONAL_FEATURES]

        pen = W_in.new_tensor(0.0)
        if ewm_idx:
            pen = pen + self.policy_ridge_lambda_base * (W_in[:, ewm_idx]**2).mean()
        if lag_idx:
            pen = pen + self.policy_ridge_lambda_lag  * (W_in[:, lag_idx]**2).mean()
        if anchor_idx:
            pen = pen + self.anchor_ridge_lambda * (W_in[:, anchor_idx]**2).mean()
        if season_idx:
            pen = pen + self.season_ridge_lambda * (W_in[:, season_idx]**2).mean()
        return pen

    def training_step(self, batch, batch_idx):
        X, y, _ = batch
        y_pred = self.model(X)
        loss_q = quantile_loss(y_pred, y, self.q_list)
        loss_pr = self._policy_ridge_penalty()
        loss = loss_q + loss_pr
        self.log("train_loss", loss, prog_bar=True, on_step=False, on_epoch=True)
        self.log("policy_ridge", loss_pr, prog_bar=False, on_step=False, on_epoch=True)
        return loss

    def validation_step(self, batch, batch_idx):
        X, y, _ = batch
        y_pred = self.model(X)
        loss_q = quantile_loss(y_pred, y, self.q_list)
        loss_pr = self._policy_ridge_penalty()
        loss = loss_q + loss_pr
        self.log("val_loss", loss, prog_bar=True, on_step=False, on_epoch=True)
        return loss

    def configure_optimizers(self):
        opt = torch.optim.Adam(self.parameters(), lr=self.lr, weight_decay=self.weight_decay)
        sch = torch.optim.lr_scheduler.ReduceLROnPlateau(opt, mode="min", patience=2, factor=0.6)
        return {"optimizer": opt, "lr_scheduler": {"scheduler": sch, "monitor": "val_loss"}}

# ---------------------------------------------------------------
# Load best params from HPO (or fallback defaults)
# ---------------------------------------------------------------
if os.path.exists(BEST_TRIAL_JSON):
    with open(BEST_TRIAL_JSON, "r") as f:
        _best = json.load(f)
    bp = _best.get("params", {})
    print("✅ Loaded best-trial params:", bp)
else:
    print("⚠️ best_trial_omicron.json not found; using conservative defaults.")
    bp = {}

# Recreate cap families from best params (fallback to suffix defaults if absent)
cap_ewm6  = float(bp.get("cap_ewm6",  DEFAULT_CAPS_BY_SUFFIX["_ewm_hl6"]))
cap_ewm12 = float(bp.get("cap_ewm12", DEFAULT_CAPS_BY_SUFFIX["_ewm_hl12"]))

POLICY_CAPS_BY_NAME_BEST = {}
for n in KNOWN_COLS:
    if n.endswith("_ewm_hl6"):    POLICY_CAPS_BY_NAME_BEST[n] = cap_ewm6
    elif n.endswith("_ewm_hl12"): POLICY_CAPS_BY_NAME_BEST[n] = cap_ewm12

POLICY_NEG_IDX_BEST, POLICY_CAPS_VEC_BEST = [], []
for i, n in enumerate(KNOWN_COLS):
    if n.startswith("covid_19_") and ("_ewm_hl" in n):
        POLICY_NEG_IDX_BEST.append(i)
        POLICY_CAPS_VEC_BEST.append(POLICY_CAPS_BY_NAME_BEST.get(n, 0.15))

# Architecture/optim (fallbacks mirror your HPO ranges/modes)
best_hparams = dict(
    horz_dim     = int(bp.get("horz_dim", 8)),
    hidden       = int(bp.get("hidden", 128)),
    dropout      = float(bp.get("dropout", 0.18)),
    lr           = float(bp.get("lr", 1.2e-3)),
    weight_decay = float(bp.get("weight_decay", 1e-5)),
    ridge_base   = float(bp.get("ridge_base", 0.02)),
    ridge_lag    = float(bp.get("ridge_lag",  0.002)),
    ridge_anchor = float(bp.get("ridge_anchor", 0.02)),
    ridge_season = float(bp.get("ridge_season", 0.3)),
)
print("✅ Final training hparams:", best_hparams)

# ---------------------------------------------------------------
# Instantiate module with best params
# ---------------------------------------------------------------
best_module = QuantileLightningModule(
    d_in=len(KNOWN_COLS),
    H=H,
    q_list=QUANTILES,
    horz_dim=best_hparams["horz_dim"],
    hidden=best_hparams["hidden"],
    dropout=best_hparams["dropout"],
    lr=best_hparams["lr"],
    weight_decay=best_hparams["weight_decay"],
    policy_ridge_lambda_base=best_hparams["ridge_base"],
    policy_ridge_lambda_lag =best_hparams["ridge_lag"],   # small L2 on lags (optional)
    anchor_ridge_lambda=best_hparams["ridge_anchor"],
    season_ridge_lambda=best_hparams["ridge_season"],
    enable_monotone_hidden=True,
    policy_feature_indices_neg=POLICY_NEG_IDX_BEST,
    policy_caps_vec=POLICY_CAPS_VEC_BEST,
)

# ---------------------------------------------------------------
# Train FINAL model (max_epochs=250, ES patience=10)
# ---------------------------------------------------------------
FINAL_CKPT = os.path.join(MODEL_DIR, "final_best_model.ckpt")
ckpt_cb = ModelCheckpoint(monitor="val_loss", mode="min", save_top_k=1, save_weights_only=True, filename="final-best")
es_cb   = EarlyStopping(monitor="val_loss", mode="min", patience=10, min_delta=1e-4)

trainer = pl.Trainer(
    max_epochs=250,
    accelerator="gpu" if torch.cuda.is_available() else "cpu",
    devices=1,
    callbacks=[ckpt_cb, es_cb],
    log_every_n_steps=50,
    enable_checkpointing=True,
    precision="16-mixed" if torch.cuda.is_available() else "32-true",
    logger=False,
    deterministic=True,
    check_val_every_n_epoch=1,
    num_sanity_val_steps=0,
    gradient_clip_val=1.0
)

trainer.fit(best_module, train_dataloaders=train_loader, val_dataloaders=val_loader)
print("✅ Training complete. Best ckpt:", ckpt_cb.best_model_path)

# Load the best weights (optional but recommended for export)
if ckpt_cb.best_model_path and os.path.isfile(ckpt_cb.best_model_path):
    state = torch.load(ckpt_cb.best_model_path, map_location="cpu")
    sd = state["state_dict"] if isinstance(state, dict) and "state_dict" in state else state
    best_module.load_state_dict(sd, strict=True)

# ---------------------------------------------------------------
# Export preprocessing artifacts (all with _omicron suffix)
# ---------------------------------------------------------------
SCHEMA_JSON   = os.path.join(MODEL_DIR, "serving_schema_omicron.json")
COUNTRY_MAP   = os.path.join(MODEL_DIR, "country_index_map_omicron.json")
VAL_COUNTRIES = os.path.join(MODEL_DIR, "val_countries_omicron.json")
CENTER_JSON   = os.path.join(MODEL_DIR, "center_means_omicron.json")

serving_notes = [
    "No PCA. Policies use RAW sliders.",
    "Policy windows are exponential weighted means (Koyck) with halflives {6,12}.",
    "Anchors: prev_ma_4w, prev_slope_4w (history-only); time_c fixed across horizon.",
    "Smoothed policy windows used directly (no residualization).",
    "Step-varying across horizon: policy windows & discrete lags; seasonality & latitude held fixed.",
    "Non-crossing quantiles via cumulative Softplus.",
    "Policy EWMs hard-constrained non-positive (per-feature caps). Lag features are unconstrained (optionally L2-regularized).",
    "Targets use stabilized logit on y/U; outputs are sigmoid-mapped logits and scaled by U at serve time.",
    "Single model; no random effects.",
    "Includes lag_gate feature; served with lag gate OFF (validation matches serving).",
]

serving_schema = {
    "time_idx": "week_id",
    "target": "covid_19_prevalence",
    "group_ids": ["country_iso3"],
    "max_prediction_length": int(H),
    "static_categoricals": ["country_iso3"],
    "static_reals": [],
    "time_varying_known_reals": list(KNOWN_COLS),
    "constants": {
        "prevalence_cap": float(TARGET_CAP_U),
    },
    "time_varying_unknown_reals": [],
    "normalization": {"type": "FeatureZNormOnly"},
    "categorical_encoders": {"country_iso3": {"encoder": "LabelMap", "add_nan": False}},
    "notes": serving_notes,
}
with open(SCHEMA_JSON, "w") as f: json.dump(serving_schema, f, indent=2)

with open(COUNTRY_MAP, "w") as f: json.dump(country_index_map, f, indent=2)
with open(VAL_COUNTRIES, "w") as f: json.dump(sorted(list(val_countries)), f, indent=2)
with open(CENTER_JSON, "w") as f: json.dump(CENTER_MEANS_FOR_INTS, f, indent=2)

# ---------------------------------------------------------------
# TorchScript export (PF-free serving)
# ---------------------------------------------------------------
SCRIPTED_PT      = os.path.join(MODEL_DIR, "model_scripted_omicron.pt")
FEATURE_CONTRACT = os.path.join(MODEL_DIR, "feature_contract_omicron.json")

best_module.eval().to("cpu")
model_for_export = best_module.model.eval().cpu()

if len(val_dataset) == 0:
    raise RuntimeError(
        "Validation set is empty after horizon construction. "
        "Try reducing H (e.g., H=8) or adjusting the train/val split."
    )

ex_X, _, _ = val_dataset[0]
ex_X = ex_X.unsqueeze(0)

class _ModelScriptWrapper(torch.nn.Module):
    def __init__(self, core_model: nn.Module, cap_u: float):
        super().__init__(); self.core = core_model; self.cap_u = float(cap_u)
    def forward(self, decoder_cont: torch.Tensor) -> torch.Tensor:
        return self.cap_u * torch.sigmoid(self.core(decoder_cont))

with torch.inference_mode():
    scripted = torch.jit.trace(_ModelScriptWrapper(model_for_export, cap_u=TARGET_CAP_U).eval(),
                               (ex_X.float(),),
                               strict=False)
scripted.save(SCRIPTED_PT)

contract = {
    "decoder_cont_lastdim": int(ex_X.shape[-1]),
    "expects": {"decoder_cont": "float32 [B, H, D_known]"},
    "produces": f"float32 [B, H, Q] in [0,{TARGET_CAP_U}] (sigmoid on logits, scaled by U)",
    "horizon": int(H),
    "quantiles": [float(q) for q in QUANTILES],
    "feature_order": list(KNOWN_COLS),
}
with open(FEATURE_CONTRACT, "w") as f: json.dump(contract, f, indent=2)

print("✅ TorchScript saved to:", SCRIPTED_PT)
print("📝 Feature contract saved to:", FEATURE_CONTRACT)

# ---------------------------------------------------------------
# Bounded-output smoke test
# ---------------------------------------------------------------
with torch.no_grad():
    logits = best_module.model(ex_X.float())     # [1,H,Q] logits
    scaled = TARGET_CAP_U * torch.sigmoid(logits)
    m = float(torch.max(scaled))
    assert m <= TARGET_CAP_U + 1e-6, f"Scaled outputs exceed U: {m} > {TARGET_CAP_U}"
print(f"✅ Bounded-output check passed (max ≤ U={TARGET_CAP_U:.6f})")

# ---------------------------------------------------------------
# Sanity checks
# ---------------------------------------------------------------
assert list(KNOWN_COLS) == serving_schema["time_varying_known_reals"], "KNOWN_COLS mismatch serve schema."

# ---------------------------------------------------------------
# Summary
# ---------------------------------------------------------------
def _mb(path): return os.path.getsize(path)/(1024*1024) if os.path.exists(path) else 0.0
print("\n✅ Export complete →", MODEL_DIR)
for p in [
    os.path.join(MODEL_DIR, "final_best_model.ckpt"),
    os.path.join(MODEL_DIR, "model_scripted_omicron.pt"),
    os.path.join(MODEL_DIR, "feature_contract_omicron.json"),
    os.path.join(MODEL_DIR, "serving_schema_omicron.json"),
    os.path.join(MODEL_DIR, "country_index_map_omicron.json"),
    os.path.join(MODEL_DIR, "val_countries_omicron.json"),
    os.path.join(MODEL_DIR, "feature_norm_stats_omicron.json"),
    os.path.join(MODEL_DIR, "center_means_omicron.json"),
    BEST_TRIAL_JSON,
]:
    if os.path.exists(p):
        print(f" - {os.path.basename(p):28s}: {_mb(p):.2f} MB")
```
